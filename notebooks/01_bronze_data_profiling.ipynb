```python
import pandas as pd

df = pd.read_json("../data/bronze/selic_raw.json")

print("Registros:", len(df))
df
```

```python
print("Valores nulos:")
print(df.isna().sum())

print("\nRegistros duplicados:")
print(df.duplicated().sum())
```

```python
print(df.dtypes)

print("\nPrimeira data:", df["data"].min())
print("Última data:", df["data"].max())

print("\nMenor valor:", df["valor"].min())
print("Maior valor:", df["valor"].max())
```

```python
df["data"] = pd.to_datetime(df["data"], format="%d/%m/%Y")
df["valor"] = pd.to_numeric(df["valor"])

print("Primeira data:", df["data"].min())
print("Última data:", df["data"].max())

print("Menor Selic:", df["valor"].min())
print("Maior Selic:", df["valor"].max())
```
